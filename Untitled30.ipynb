```python
from rdkit import Chem
import pandas as pd

# Load B3DB
b3db_path = "B3DB/B3DB/B3DB_classification_extended.tsv.gz"

b3db = pd.read_csv(
    b3db_path,
    sep="\t",
    compression="gzip",
    low_memory=False
)

# Keep only the columns required for our project
b3db_small = b3db[["SMILES", "BBB+/BBB-"]].copy()

# Remove missing values
b3db_small = b3db_small.dropna(
    subset=["SMILES", "BBB+/BBB-"]
)

# Canonicalize SMILES
def canonicalize_smiles(smiles):
    mol = Chem.MolFromSmiles(str(smiles))
    if mol is None:
        return None
    return Chem.MolToSmiles(mol, canonical=True)

b3db_small["canonical_smiles"] = (
    b3db_small["SMILES"].apply(canonicalize_smiles)
)

b3db_small = b3db_small.dropna(
    subset=["canonical_smiles"]
)

# Remove duplicate canonical SMILES
b3db_small = b3db_small.drop_duplicates(
    subset=["canonical_smiles"]
)

# Canonicalize BACE1 SMILES
bace1_clean["canonical_smiles"] = (
    bace1_clean["canonical_smiles"].apply(canonicalize_smiles)
)

bace1_clean = bace1_clean.dropna(
    subset=["canonical_smiles"]
)

# Match BACE1 molecules with B3DB
bace1_bbb = bace1_clean.merge(
    b3db_small[["canonical_smiles", "BBB+/BBB-"]],
    on="canonical_smiles",
    how="inner"
)

print("BACE1 molecules:", len(bace1_clean))
print("B3DB molecules:", len(b3db_small))
print("Matched BACE1 + BBB molecules:", len(bace1_bbb))

print("\nBBB distribution:")
print(bace1_bbb["BBB+/BBB-"].value_counts())
```

```python
# ============================================================
# CORRECTED BACE1–B3DB STRUCTURE MATCHING
# ============================================================

from rdkit import Chem
import pandas as pd

def standardize_smiles(smiles):
    """
    Standardize a SMILES for structure matching.

    Steps:
    1. Parse molecule with RDKit
    2. Split disconnected fragments
    3. Keep the largest fragment by heavy-atom count
    4. Remove stereochemical information for initial matching
    5. Return canonical SMILES
    """
    try:
        mol = Chem.MolFromSmiles(str(smiles))

        if mol is None:
            return None

        fragments = Chem.GetMolFrags(
            mol,
            asMols=True,
            sanitizeFrags=True
        )

        if len(fragments) == 0:
            return None

        # Largest fragment = most heavy atoms
        mol = max(
            fragments,
            key=lambda m: m.GetNumHeavyAtoms()
        )

        # Remove stereochemistry only for matching
        Chem.RemoveStereochemistry(mol)

        return Chem.MolToSmiles(
            mol,
            canonical=True
        )

    except Exception:
        return None


# ------------------------------------------------------------
# BACE1
# ------------------------------------------------------------

bace1_match = bace1_clean.copy()

bace1_match["match_smiles"] = bace1_match[
    "canonical_smiles"
].apply(standardize_smiles)

bace1_match = bace1_match.dropna(
    subset=["match_smiles"]
)

bace1_match = bace1_match.drop_duplicates(
    subset=["match_smiles"]
)


# ------------------------------------------------------------
# B3DB
# ------------------------------------------------------------

b3db_match = b3db_small.copy()

b3db_match["match_smiles"] = b3db_match[
    "canonical_smiles"
].apply(standardize_smiles)

b3db_match = b3db_match.dropna(
    subset=["match_smiles"]
)

b3db_match = b3db_match.drop_duplicates(
    subset=["match_smiles"]
)


# ------------------------------------------------------------
# MATCH
# ------------------------------------------------------------

bace1_bbb = bace1_match.merge(
    b3db_match[
        ["match_smiles", "BBB+/BBB-"]
    ],
    on="match_smiles",
    how="inner"
)


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("BACE1 standardized molecules:", len(bace1_match))
print("B3DB standardized molecules:", len(b3db_match))
print("Matched BACE1 + BBB molecules:", len(bace1_bbb))

print("\nBBB distribution:")
print(
    bace1_bbb["BBB+/BBB-"].value_counts()
)

print("\nFirst matched molecules:")
display(
    bace1_bbb[
        [
            "match_smiles",
            "BBB+/BBB-"
        ]
    ].head(20)
)
```

```python
# ============================================================
# BBB REGRESSION DATASET PREPARATION
# Source: B3DB
# Target: experimental quantitative logBB
# Purpose: prepare clean data for BBB regression
# ============================================================

import pandas as pd
import numpy as np

# Keep only the fields required for regression
bbb_reg = b3db[["SMILES", "logBB"]].copy()

# Convert logBB to numeric
bbb_reg["logBB"] = pd.to_numeric(
    bbb_reg["logBB"],
    errors="coerce"
)

# Remove missing SMILES and missing/non-numeric logBB
bbb_reg = bbb_reg.dropna(
    subset=["SMILES", "logBB"]
).copy()

# Remove exact duplicate rows
bbb_reg = bbb_reg.drop_duplicates(
    subset=["SMILES", "logBB"]
).reset_index(drop=True)

print("==============================================")
print("B3DB BBB REGRESSION DATASET")
print("==============================================")
print("Records:", len(bbb_reg))
print("Missing SMILES:", bbb_reg["SMILES"].isna().sum())
print("Missing logBB:", bbb_reg["logBB"].isna().sum())
print("\nlogBB statistics:")
print(bbb_reg["logBB"].describe())
```

```python
# ============================================================
# CHECK BBB REGRESSION DATA QUALITY
# ============================================================

from rdkit import Chem

def is_valid_smiles(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles)
        return mol is not None
    except:
        return False

bbb_reg["valid_smiles"] = bbb_reg["SMILES"].apply(is_valid_smiles)

print("Total records:", len(bbb_reg))
print("Valid SMILES:", bbb_reg["valid_smiles"].sum())
print("Invalid SMILES:", (~bbb_reg["valid_smiles"]).sum())

# Keep only valid molecules
bbb_reg = bbb_reg[
    bbb_reg["valid_smiles"]
].drop(columns=["valid_smiles"]).reset_index(drop=True)

print("\nFinal BBB regression records:", len(bbb_reg))
print("\nFinal logBB distribution:")
print(bbb_reg["logBB"].describe())
```

```python
# ============================================================
# BBB REGRESSION — MOLECULAR REPRESENTATION
# Source: B3DB
# Target: experimental logBB
# Representation: Morgan fingerprints
# ============================================================

from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np

# Convert SMILES to RDKit molecules
bbb_reg["mol"] = bbb_reg["SMILES"].apply(Chem.MolFromSmiles)

# Generate Morgan fingerprints
# radius=2 corresponds to ECFP4-style local molecular features
# nBits=2048 provides a fixed-length molecular representation
def get_morgan_fp(mol):
    return np.array(
        AllChem.GetMorganFingerprintAsBitVect(
            mol,
            radius=2,
            nBits=2048
        )
    )

X = np.array([
    get_morgan_fp(mol)
    for mol in bbb_reg["mol"]
])

y = bbb_reg["logBB"].values.astype(np.float32)

print("==============================================")
print("BBB REGRESSION FEATURES")
print("==============================================")
print("Number of molecules:", len(X))
print("Fingerprint shape:", X.shape)
print("Target shape:", y.shape)
print("Target: experimental logBB")
print("Fingerprint: Morgan radius=2, 2048 bits")
```

```python
# ============================================================
# BBB REGRESSION — SCAFFOLD-BASED TRAIN/VALIDATION/TEST SPLIT
# Source: B3DB
# Split: Bemis-Murcko molecular scaffolds
# Purpose: reduce chemical scaffold leakage
# ============================================================

from rdkit.Chem.Scaffolds import MurckoScaffold
from collections import defaultdict
import numpy as np

# ------------------------------------------------------------
# 1. Generate Bemis-Murcko scaffold for every molecule
# ------------------------------------------------------------

def get_scaffold(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None

    return MurckoScaffold.MurckoScaffoldSmiles(
        mol=mol,
        includeChirality=False
    )

bbb_reg["scaffold"] = bbb_reg["SMILES"].apply(get_scaffold)

# Remove molecules for which a scaffold could not be generated
bbb_reg = bbb_reg.dropna(subset=["scaffold"]).reset_index(drop=True)

# ------------------------------------------------------------
# 2. Group molecules by scaffold
# ------------------------------------------------------------

scaffold_groups = defaultdict(list)

for idx, scaffold in enumerate(bbb_reg["scaffold"]):
    scaffold_groups[scaffold].append(idx)

scaffold_list = list(scaffold_groups.keys())

# Reproducible shuffle
rng = np.random.default_rng(42)
rng.shuffle(scaffold_list)

# ------------------------------------------------------------
# 3. Assign complete scaffolds to train / validation / test
# ------------------------------------------------------------

train_target = 0.70 * len(bbb_reg)
valid_target = 0.15 * len(bbb_reg)

train_idx = []
valid_idx = []
test_idx = []

for scaffold in scaffold_list:

    indices = scaffold_groups[scaffold]

    if len(train_idx) < train_target:
        train_idx.extend(indices)

    elif len(valid_idx) < valid_target:
        valid_idx.extend(indices)

    else:
        test_idx.extend(indices)

# Convert to arrays
train_idx = np.array(train_idx)
valid_idx = np.array(valid_idx)
test_idx = np.array(test_idx)

# ------------------------------------------------------------
# 4. Create feature/target splits
# ------------------------------------------------------------

X_train = X[train_idx]
y_train = y[train_idx]

X_valid = X[valid_idx]
y_valid = y[valid_idx]

X_test = X[test_idx]
y_test = y[test_idx]

# ------------------------------------------------------------
# 5. Verify scaffold separation
# ------------------------------------------------------------

train_scaffolds = set(bbb_reg.iloc[train_idx]["scaffold"])
valid_scaffolds = set(bbb_reg.iloc[valid_idx]["scaffold"])
test_scaffolds = set(bbb_reg.iloc[test_idx]["scaffold"])

print("==============================================")
print("BBB REGRESSION SCAFFOLD SPLIT")
print("==============================================")

print("\nMolecule counts:")
print("Total:", len(bbb_reg))
print("Train:", len(train_idx))
print("Validation:", len(valid_idx))
print("Test:", len(test_idx))

print("\nScaffold counts:")
print("Train:", len(train_scaffolds))
print("Validation:", len(valid_scaffolds))
print("Test:", len(test_scaffolds))

print("\nScaffold overlap:")
print("Train ∩ Validation:", len(train_scaffolds & valid_scaffolds))
print("Train ∩ Test:", len(train_scaffolds & test_scaffolds))
print("Validation ∩ Test:", len(valid_scaffolds & test_scaffolds))

print("\nTarget means:")
print("Train logBB mean:", y_train.mean())
print("Validation logBB mean:", y_valid.mean())
print("Test logBB mean:", y_test.mean())
```

```python
# ============================================================
# BBB REGRESSION — RANDOM FOREST
# Source: B3DB
# Input: Morgan fingerprints (2048 bits)
# Target: experimental logBB
# Split: scaffold-based
# ============================================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# ------------------------------------------------------------
# Model configuration
# ------------------------------------------------------------

rf_model = RandomForestRegressor(
    n_estimators=500,
    max_features="sqrt",
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

print("==============================================")
print("TRAINING BBB REGRESSION MODEL")
print("==============================================")
print("Model: Random Forest Regressor")
print("Trees: 500")
print("Features: Morgan fingerprint, 2048 bits")
print("Training samples:", len(X_train))
print("Validation samples:", len(X_valid))
print("Test samples:", len(X_test))

# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

rf_model.fit(X_train, y_train)

print("\nTraining completed.")

# ------------------------------------------------------------
# Validation prediction
# ------------------------------------------------------------

y_valid_pred = rf_model.predict(X_valid)

valid_mae = mean_absolute_error(y_valid, y_valid_pred)
valid_rmse = np.sqrt(mean_squared_error(y_valid, y_valid_pred))
valid_r2 = r2_score(y_valid, y_valid_pred)

print("\n==============================================")
print("VALIDATION RESULTS")
print("==============================================")
print(f"MAE  : {valid_mae:.4f}")
print(f"RMSE : {valid_rmse:.4f}")
print(f"R²   : {valid_r2:.4f}")
```

```python
# ============================================================
# BBB REGRESSION — EXTRA TREES
# Source: B3DB
# Input: Morgan fingerprints (2048 bits)
# Target: experimental logBB
# Split: same scaffold-based split as Random Forest
# ============================================================

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

extra_model = ExtraTreesRegressor(
    n_estimators=500,
    max_features="sqrt",
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

print("==============================================")
print("TRAINING BBB EXTRA TREES REGRESSOR")
print("==============================================")
print("Model: Extra Trees Regressor")
print("Trees: 500")
print("Features: Morgan fingerprint, 2048 bits")
print("Training samples:", len(X_train))
print("Validation samples:", len(X_valid))

extra_model.fit(X_train, y_train)

print("\nTraining completed.")

# Validation prediction
y_valid_pred_extra = extra_model.predict(X_valid)

extra_valid_mae = mean_absolute_error(
    y_valid,
    y_valid_pred_extra
)

extra_valid_rmse = np.sqrt(
    mean_squared_error(
        y_valid,
        y_valid_pred_extra
    )
)

extra_valid_r2 = r2_score(
    y_valid,
    y_valid_pred_extra
)

print("\n==============================================")
print("EXTRA TREES VALIDATION RESULTS")
print("==============================================")
print(f"MAE  : {extra_valid_mae:.4f}")
print(f"RMSE : {extra_valid_rmse:.4f}")
print(f"R²   : {extra_valid_r2:.4f}")
```

```python
# ============================================================
# BBB REGRESSION — FINAL TEST EVALUATION
# Model: Extra Trees Regressor
# Test set: completely untouched scaffold split
# ============================================================

# Predict on the test set
y_test_pred = extra_model.predict(X_test)

# Calculate final test metrics
test_mae = mean_absolute_error(
    y_test,
    y_test_pred
)

test_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred
    )
)

test_r2 = r2_score(
    y_test,
    y_test_pred
)

print("==============================================")
print("FINAL BBB REGRESSION TEST RESULTS")
print("==============================================")
print("Model: Extra Trees Regressor")
print("Test samples:", len(X_test))
print()
print(f"MAE  : {test_mae:.4f}")
print(f"RMSE : {test_rmse:.4f}")
print(f"R²   : {test_r2:.4f}")
```

```python
"""
Continuous BBB-conditioned DrugGEN dataset.

Condition:
    [target_condition, normalized_logBB]

For this project:
    target_condition = 1.0  -> BACE1
    normalized_logBB        -> continuous BBB permeability signal

The BBB signal comes from the B3DB-trained Extra Trees regression model.
"""

import pandas as pd
import torch
from torch.utils.data import Dataset
from rdkit import Chem


def canonical(smiles):
    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return None

    return Chem.MolToSmiles(mol)


class ConditionedDruggenDataset(Dataset):

    def __init__(self, base_dataset, labels_csv):

        labels = pd.read_csv(labels_csv)

        required = {
            "canonical_smiles",
            "bbb_condition",
            "target_condition"
        }

        missing = required - set(labels.columns)

        if missing:
            raise ValueError(
                f"Missing required columns: {sorted(missing)}"
            )

        labels["canonical_smiles"] = (
            labels["canonical_smiles"].map(canonical)
        )

        labels = labels.dropna(
            subset=["canonical_smiles"]
        )

        labels = labels.drop_duplicates(
            subset=["canonical_smiles"]
        )

        self.labels = {}

        for _, row in labels.iterrows():

            self.labels[row["canonical_smiles"]] = {
                "target": float(row["target_condition"]),
                "bbb": float(row["bbb_condition"])
            }

        self.base = base_dataset

        missing_count = 0

        for i in range(len(base_dataset)):

            smiles = base_dataset[i].smiles
            key = canonical(smiles)

            if key not in self.labels:
                missing_count += 1

        if missing_count > 0:

            raise ValueError(
                f"{missing_count} target molecules have no "
                "continuous BBB condition."
            )

    def __len__(self):
        return len(self.base)

    def __getitem__(self, idx):

        data = self.base[idx].clone()

        key = canonical(data.smiles)

        values = self.labels[key]

        data.condition = torch.tensor(
            [
                values["target"],
                values["bbb"]
            ],
            dtype=torch.float32
        )

        return data
```

```python
"""
Continuous-condition extension of DrugGEN.

Original DrugGEN architecture is retained:
    Node MLP
    Edge MLP
    Transformer Encoder
    Readout layers

Added conditioning:
    [BACE1 target condition, normalized BBB logBB]

BBB is continuous, not binary.
"""

import torch
import torch.nn as nn

from src.model.layers import TransformerEncoder


def _activation(name):

    return {
        "relu": nn.ReLU(),
        "leaky": nn.LeakyReLU(),
        "sigmoid": nn.Sigmoid(),
        "tanh": nn.Tanh()
    }[name]


class ConditionalGenerator(nn.Module):

    def __init__(
        self,
        act,
        vertexes,
        edges,
        nodes,
        dropout,
        dim,
        depth,
        heads,
        mlp_ratio,
        condition_dim=2
    ):

        super().__init__()

        self.vertexes = vertexes
        self.edges = edges
        self.nodes = nodes
        self.dim = dim

        activation = _activation(act)

        # Original DrugGEN node pathway
        self.node_layers = nn.Sequential(
            nn.Linear(nodes, 64),
            activation,
            nn.Linear(64, dim),
            activation,
            nn.Dropout(dropout)
        )

        # Original DrugGEN edge pathway
        self.edge_layers = nn.Sequential(
            nn.Linear(edges, 64),
            activation,
            nn.Linear(64, dim),
            activation,
            nn.Dropout(dropout)
        )

        # New continuous condition pathway
        self.condition_layers = nn.Sequential(
            nn.Linear(condition_dim, dim),
            nn.Tanh()
        )

        # Original Transformer
        self.TransformerEncoder = TransformerEncoder(
            dim=dim,
            depth=depth,
            heads=heads,
            act=activation,
            mlp_ratio=mlp_ratio,
            drop_rate=dropout
        )

        # Original readouts
        self.readout_n = nn.Linear(dim, nodes)
        self.readout_e = nn.Linear(dim, edges)

    def forward(self, z_e, z_n, condition):

        node = self.node_layers(z_n)
        edge = self.edge_layers(z_e)

        # condition:
        # [target_condition, normalized_logBB]
        c = self.condition_layers(condition)

        node = node + c.unsqueeze(1)

        edge = edge + c.unsqueeze(1).unsqueeze(1)

        # Preserve symmetric molecular graph
        edge = (
            edge +
            edge.permute(0, 2, 1, 3)
        ) / 2

        node, edge = self.TransformerEncoder(
            node,
            edge
        )

        node_output = self.readout_n(node)
        edge_output = self.readout_e(edge)

        return (
            node,
            edge,
            node_output,
            edge_output
        )


class ConditionalDiscriminator(nn.Module):

    def __init__(
        self,
        act,
        vertexes,
        edges,
        nodes,
        dropout,
        dim,
        depth,
        heads,
        mlp_ratio,
        condition_dim=2
    ):

        super().__init__()

        self.vertexes = vertexes
        self.dim = dim

        activation = _activation(act)

        self.node_layers = nn.Sequential(
            nn.Linear(nodes, 64),
            activation,
            nn.Linear(64, dim),
            activation,
            nn.Dropout(dropout)
        )

        self.edge_layers = nn.Sequential(
            nn.Linear(edges, 64),
            activation,
            nn.Linear(64, dim),
            activation,
            nn.Dropout(dropout)
        )

        # Continuous target + BBB conditioning
        self.condition_layers = nn.Sequential(
            nn.Linear(condition_dim, dim),
            nn.Tanh()
        )

        self.TransformerEncoder = TransformerEncoder(
            dim=dim,
            depth=depth,
            heads=heads,
            act=activation,
            mlp_ratio=mlp_ratio,
            drop_rate=dropout
        )

        self.node_mlp = nn.Sequential(
            nn.Linear(vertexes * dim, 64),
            activation,
            nn.Linear(64, 32),
            activation,
            nn.Linear(32, 16),
            activation,
            nn.Linear(16, 1)
        )

    def forward(self, z_e, z_n, condition):

        batch_size = z_n.shape[0]

        node = self.node_layers(z_n)
        edge = self.edge_layers(z_e)

        c = self.condition_layers(condition)

        node = node + c.unsqueeze(1)

        edge = edge + c.unsqueeze(1).unsqueeze(1)

        edge = (
            edge +
            edge.permute(0, 2, 1, 3)
        ) / 2

        node, _ = self.TransformerEncoder(
            node,
            edge
        )

        node = node.reshape(
            batch_size,
            -1
        )

        return self.node_mlp(node)
```
